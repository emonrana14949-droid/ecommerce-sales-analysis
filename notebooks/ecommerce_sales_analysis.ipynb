# E-commerce Sales Analysis

A Pandas, NumPy and Matplotlib walkthrough of the e-commerce sales dataset.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
df = pd.read_csv('../data/ecommerce_sales_dataset.csv', parse_dates=['order_date'])
df.head()
```

## 1. Revenue by product category


```python
by_category = df.groupby('product_category').agg(
    orders=('order_id', 'count'),
    total_revenue_bdt=('revenue_bdt', 'sum'),
    avg_discount_percentage=('discount_percentage', 'mean')
).round(2).sort_values('total_revenue_bdt', ascending=False)
by_category
```

## 2. Monthly revenue trend


```python
monthly = df.groupby('month').agg(
    orders=('order_id', 'count'),
    total_revenue_bdt=('revenue_bdt', 'sum')
)

plt.figure(figsize=(9, 4.8))
plt.plot(monthly.index, monthly['total_revenue_bdt'], marker='o', linewidth=2)
plt.title('Monthly Revenue Trend — 2025')
plt.xlabel('Month')
plt.ylabel('Total revenue (BDT)')
plt.show()
```
